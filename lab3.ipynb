```python
# BFS = Breadth-First Search

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': [],
    'G': []
}

def BFS(start):
    visited = []
    queue = [start]

    while queue:
        node = queue.pop(0)

        if node not in visited:
            print(node, end=" ")
            visited.append(node)

            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)

print("BFS Traversal:")
BFS('A')
```

```python
graph = {
    0: [],
    1: [0],
    2: [10, 6],
    3: [1, 2, 4, 7],
    4: [5, 8, 12],
    5: [],
    6: [],
    7: [9],
    8: [],
    9: [],
    10: [11],
    11: [],
    12: [13],
    13: []
}

def DFS(start, visited=None):
    if visited is None:
        visited = []
    if start not in visited:
        print(start, end=" ")
        visited.append(start)
        for neighbor in graph[start]:
            DFS(neighbor, visited)

print("DFS Traversal:")
DFS(3)
```
